Manipulação de Dados Tabulares com Pandas
Funções Básicas

```python
import pandas as pd

cidades = pd.DataFrame(
    [
        ['Paraná', 'Londrina', 575377, 1356.00],
        ['São Paulo', 'São Carlos', 254484, 1508.00],
        ['Santa Catarina', 'Florianópolis', 508826, 1798.00],
        ['Paraná', 'Curitiba', 1963726, 2293.00],
        ['São Paulo', 'Campinas', 1223237, 1710.00]
    ],
    columns=['Estado', 'Cidade', 'Habitantes', 'Salário-Médio']
)

cidades
```

```python
cidades.head(2)
cidades.tail(2)
cidades.shape
cidades.info()
cidades.columns
```

Tipos de Dados com Pandas

```python
cidades['Salário-Médio'].describe()
cidades.dtypes
cidades.info()
cidades['Habitantes'] = cidades['Habitantes'].astype('float64')

cidades.dtypes
```

Trabalhando com datas

```python
cidades['Data'] = [
    '3-8-2021',
    '3-9-2021',
    '3-10-2021',
    '3-11-2021',
    '3-12-2021'
]

cidades['Data'] = pd.to_datetime(
    cidades['Data'],
    format='%d-%m-%Y'
)

cidades

cidades['Data'] = cidades['Data'].dt.strftime('%d-%m-%Y')

cidades
```

Renomeando colunas

```python
cidades.rename(
    columns={
        'Habitantes': 'N Habitantes',
        'Data': 'Data - Censo'
    },
    inplace=True
)

cidades
```

Selecionando linhas e colunas

```python
cidadesHab = cidades[['N Habitantes', 'Salário-Médio']]

cidadesHab

cidades.loc[[1, 2]]

cidades.iloc[1:3]

cidades.iloc[:, 0:2]

cidades.iloc[1:3, :3]

cidades.iloc[:, [1, 3, 0, 2]]
```

Adicionando e removendo colunas

```python
cidades.insert(
    2,
    'Sigla',
    ['PR', 'SP', 'SC', 'PR', 'SP']
)

cidades
```

```python
cidades.insert(
    2,
    'Sigla-Repetida',
    cidades['Sigla']
)

cidades
```

```python
cidades = cidades.drop(
    ['Sigla-Repetida'],
    axis=1
)

cidades
```

Operações Básicas - Consultas

```python
cidades.query('`Salário-Médio` > 1400.0')
```

```python
cidades.query('Estado == "São Paulo"')
```

```python
cidades.query(
    '`N Habitantes` > 500000 and `N Habitantes` < 1300000'
)
```

Operações Básicas - Ordenação

```python
cidades.sort_values(
    by='Cidade',
    inplace=True
)

cidades
```

```python
cidades.sort_values(
    by='Salário-Médio',
    inplace=True
)

cidades
```

Operações Básicas - Combinando DataFrames

```python
cidades_salarios = pd.DataFrame(
    [
        ['Paraná', 'Londrina', 1356.00],
        ['São Paulo', 'São Carlos', 1508.00],
        ['Santa Catarina', 'Florianópolis', 1798.00],
        ['Paraná', 'Curitiba', 2293.00],
        ['São Paulo', 'Campinas', 1710.00]
    ],
    columns=['Estado', 'Cidade', 'Salário-Médio']
)

cidades_salarios
```

```python
cidades_pop = pd.DataFrame(
    [
        ['Londrina', 575377],
        ['São Carlos', 254484],
        ['Florianópolis', 508826],
        ['Curitiba', 1963726],
        ['Campinas', 1223237]
    ],
    columns=['Cidade', 'Habitantes']
)

cidades_pop
```

```python
pd.merge(
    left=cidades_salarios,
    right=cidades_pop,
    on="Cidade"
)
```

```python
pd.merge(
    left=cidades_salarios,
    right=cidades_pop,
    on="Cidade",
    how="outer"
)
```

Operações Básicas - Salvando DataFrames

```python
cidades_df1 = pd.DataFrame(
    [
        ['Londrina', 575377],
        ['São Carlos', 254484],
        ['Florianópolis', 508826]
    ],
    columns=['Cidade', 'Habitantes']
)

cidades_df1
```

```python
cidades_df2 = pd.DataFrame(
    [
        ['Curitiba', 1963726],
        ['Campinas', 1223237]
    ],
    columns=['Cidade', 'Habitantes']
)

cidades_df2
```

```python
pd.concat(
    [cidades_df1, cidades_df2],
    ignore_index=True
)
```
